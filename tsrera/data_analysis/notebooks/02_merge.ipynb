# Merge: listings ↔ RERA projects\n\nNo shared ID exists between the two sources. `entity_match.py` (imported below) fuzzy-matches listing `Society` against RERA `Project Name`, pincode-blocked where possible (threshold 85), falling back to a name-only match across all RERA projects (threshold 90) otherwise. Every row keeps `match_score` / `match_method` so low-confidence links stay visible.

```python
import sys
sys.path.insert(0, "../scripts")
import pandas as pd
from entity_match import match_listings_to_rera

RAW = "../raw"
listings = pd.read_csv(f"{RAW}/All_Merged_Updated.csv", low_memory=False)
listings.columns = [c.strip().lstrip("\ufeff") for c in listings.columns]
rera = pd.read_csv(f"{RAW}/projects_from_jsonl.csv", low_memory=False)
listings.shape, rera.shape
```

```python
merged, report = match_listings_to_rera(listings, rera)
report
```

```python
import matplotlib.pyplot as plt
```

```python
merged["match_method"].value_counts().plot.barh(figsize=(6,3), color="#2a78d6")
plt.title("Match method breakdown")
plt.gca().invert_yaxis()
plt.tight_layout()
```

```python
out_path = f"{RAW}/merged_listings_rera.csv"
merged.to_csv(out_path, index=False)
print(f"wrote {len(merged):,} rows -> {out_path}")
```
